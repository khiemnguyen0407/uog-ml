```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
# Define a neural network
# Each subdomain has its own local neural network

class MLP(nn.Module):

    def __init__(self, hidden_size=16):

        super().__init__()

        self.net = nn.Sequential(nn.Linear(1, hidden_size),
                                 nn.Tanh(),
                                 nn.Linear(hidden_size, hidden_size),
                                 nn.Tanh(),
                                 nn.Linear(hidden_size, 1)
                                 )
```
